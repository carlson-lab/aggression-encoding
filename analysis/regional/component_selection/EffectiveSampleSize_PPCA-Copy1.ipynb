```python
def autocorr3(x):
    coeffs = 1
    for i in range(1,10):
        coeffs += 2*np.dot(x[i:],x[:-1*i])/len(x)
    return coeffs
```

```python
nn,pp = X.shape
corrs = np.zeros(pp)
for i in range(pp):
    aa = autocorr3(X[:,i])
    corrs[i] = aa
corrs = corrs
```
