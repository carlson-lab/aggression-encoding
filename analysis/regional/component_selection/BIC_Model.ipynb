# Exploratory latent-dimension selection

This notebook computes the criterion implemented below: `i * p * log(N) + avg_loss * N * p`, using saved reconstruction errors. The parameter-count term is `i * p`; the loss term assumes a unit-variance Gaussian reconstruction model, up to constants. It is an exploratory generative-model calculation, not by itself an implementation of the complete supervised-model selection procedure described in the manuscript.

```python
import numpy as np
import cloudpickle
import matplotlib.pyplot as plt
```

```python
N = 2781
p = 9856
recon_losses_test = np.genfromtxt('NFM_recon_test.txt')
```

```python
def BIC(N,p,i,avg_loss):
    complexity = i*p*np.log(N)
    likelihood = -avg_loss/2
    bic = complexity - 2*likelihood*N*p
    return bic
```

```python
component_range = np.arange(20) + 1
```

```python
bics = np.zeros(20)

for i in range(20):
    bics[i] = BIC(N,p,component_range[i],recon_losses_test[i])
```

```python
nn = 10**6
plt.plot(component_range,bics/nn,lw=4,c='dodgerblue',alpha=.8)
plt.xlabel('Number of components',fontsize=20)
plt.ylabel('BIC (10^6)',fontsize=20)
plt.title('Complexity vs Dimensionality',fontsize=22)
plt.xlim([0.8,20.2])
plt.ylim([5.35*10**6/nn,6.3*10**6/nn])
plt.xticks(ticks=[1.0,3.0,5.0,7,9,11.0,13,15.0,17,19])
plt.yticks(ticks=[5.4*10**6/nn,5.7*10**6/nn,6.0*10**6/nn])
plt.plot(7, np.amin(bics)/nn, marker="o", markersize=15, markeredgecolor="k", 
         markerfacecolor="k")
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.savefig('BIC_Dimension_Selection.pdf')
```
