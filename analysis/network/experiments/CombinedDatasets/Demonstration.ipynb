```python
data_file_name = 'Your_data_file.mat'
power,coherence,granger,labels = load_data(data_file_name,fBounds=(1,56),
                        feature_list=['power','coherence','granger'])
```

```python
pf = np.squeeze(labels['powerFeatures'])
cf = np.squeeze(labels['cohFeatures'])
gf = np.squeeze(labels['gcFeatures'])

all_features = np.concatenate((pf,cf,gf))
```

## Method 1 from saved model

```python
myDict = pickle.load(open('nmf_linear_ri_final.pkl','rb'))
model = myDict['model']
components = model.components_
```

## Method 2 from text file

```python
# components = np.genfromtxt('nmf_linear_final_components.txt',delimiter=',')
```

```python
## Square components
comp2 = components**2

## Divide each feature by sum over all components "Relative strength"
comp2s = np.sum(comp2,axis=0)

components_normalized = comp2/comp2s
```

## Now save as output mat file

```python
AUSTIN_IS_GOAT = {'component_supervised':components_normalized[0],
            'components_unsupervised':components_normalized[1:],
                  'ukuNorm':components_normalized,
                  'Freqs':,np.arange(1,57),
                  'brainRegions':,labels['area'],
            'features':all_features}
from scipy.io import savemat
myNetworkName = 'Reward_network.mat'
savemat(myNetworkName,AUSTIN_IS_GOAT)
```
